Dans la prochaine étape, on va regarder du code Python qui nous permettra de créer des graphes et, ensuite, on va implémenter ensemble toutes les étapes nécessaires pour l'algo glouton vu dans le chapitre 3.3

````python
# c'est la librairie que l'on va utiliser pour créer notre graphe
import networkx as nx

# on crée un objet Graph
G = nx.Graph()

# avec les fonctions add_node et add_edge, on va créer des sommets et des arêtes
G.add_node( 1 )
G.add_node( 2 )
G.add_node( 3 )
G.add_node( 4 )
G.add_node( 5 )

G.add_edge( 1, 3 )
G.add_edge( 1, 4 )
G.add_edge( 2, 4 )
G.add_edge( 2, 5 )
G.add_edge( 3, 5 )

# le dictionnaire suivant définit des positions pour chaque sommet (où on va a dessiner le sommet sur un cahier carré)
# un dictionnaire est une structure de données qui nous permet de stocker des choses sous la forme  clé : valeur
# par example, vous pouvez stocker
pos = {1:(0,0), 2:(1,-0.8), 3:(0.7,-2.5), 4:(-0.7,-2.5), 5:(-1,-0.8)}

# cet outil nous permet d'afficher un graphe
nx.draw(G,
	with_labels=True,
  pos=pos,
	node_size=1600)
````

### Vérifier une coloration

On a bien appris comment créer un graphe et comment le colorier pour visualiser les couleurs. \
Maintenant on passe à la vérification d'une coloration : avec le dessin nous sommes capables de voir et de comprendre si le graphe est correctement colorié en vérifiant la seule contrainte du coloriage: **deux sommets voisins n'ont jamais la même couleur**.

Est-ce qu'on pourrait créer une fonction qui vérifie ça de manière automatique?

A nouveau, on va apprendre quelques petits trucs

````python
# specifier notre couleurs
colors = { 1: "blue", 2: "red", 3:"blue", 4:"yellow", 5:"green" }

# for (pour toutes les valeurs) cycle: cet outil nous permet de boucler sur toutes les différentes arêtes de notre graphe
for arret in list(G.edges()):
  # cette fonction il fait rien du tout :)
  pass


# if-else (si condition alors chose1 sinon chose2) test: cet outil nous permet de tester une condition : par exemple, je pourrais tester la chose suivante
# pour toutes arêtes de mon graphe, je teste si l'arête actuelle a l'un des deux sommets égal à 1: si oui, alors j'imprime sur l'écran le mot "oui", sinon j'imprime sur l'écran le mot "non"

# d'abord, je fais une boucle sur toutes les arêtes possibles
for arret in list(G.edges()):
  # j'imprime l'arête
  print(arret)
  # les deux commandes suivantes me permettent de récupérer les sommets correspondant à l'arête
  premier_sommet = arret[0]
  deuxieme_sommet = arret[1]
  # je regarde si le premier sommet est 1 ou (or dans Python) le deuxième sommet est 1
  if (premier_sommet==1 or deuxieme_sommet==1):
    print("oui")
  else:
    print("non")

  # de la meme manière, je peux aussi récupérer le couleur des sommets correspondants avec
  couleur_premier_sommet = colors[premier_sommet]
  couleur_deuxieme_sommet = colors[deuxieme_sommet]
````

On voit bien que l'on affiche "oui" quand on considère les arêtes (1,3) et (1,4), et "non" dans les autres cas

Quelles sont les étapes à poursuivre pour faire le coloriage de manière glouton?

1. Faire une boucle sur les sommets du graphe
2. Pour chaque sommet, on regarde le première couleur disponible dans notre liste de couleurs (disons rouge)
3. On fait une boucle sur toutes les arêtes: s'il n'y a pas des sommets voisins avec la couleur _rouge_, alors on peut colorier le sommet actuel en rouge
4. Si on trouve un voisin déjà colorié en rouge, alors on passe à la couleur suivante
